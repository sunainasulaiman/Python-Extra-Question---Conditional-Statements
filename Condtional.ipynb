## Simple Calculator

```python
num1=int(input("Enter the first number : "))
num2=int(input("Enter the second number :"))
op=input("Enter the operator")
if op == "+" :
    ans=num1+num2
    print("Answer is ",ans)
elif op == "-" :
    ans=num1-num2
    print("Answer is ",ans)
elif op == "*" :
    ans=num1*num2
    print("Answer is ",ans)
elif op == "/" :
    ans=num1/num2
    print("Answer is ",ans)
else :
    print("Invalid operator !!!")
```

```python
#1.	Write a python program to sum up all the items in a list.
num=[1,2,3,4,5]
l=len(num)
i=0
total=0
while i<l:
    total+=num[i]
    i+=1
    
print("Sum is :",total)
```

```python
#2.	Write a python program to get the largest number from a list.
num_list=[4,8,3,9,1]
l=len(num_list)
i=1
large=num_list[0]
while i<l:
    if num_list[i]>large:
        large=num_list[i]
    i+=1
print("Largets number is :",large)
```

```python
#3.	Write a python program to count the occurrence of an element in a list.
num_list = [100, 420, 100, 330, 100, 40]

n = int(input("Enter element to count: "))

count = 0
i = 0

while i < len(num_list):
    if num_list[i] == n:
        count = count + 1

    i = i + 1

print("Occurrence:", count)
```

```python
#4.	Write a python program to sum the number of integers in a list.

items = [10, "Data", 20, 30, "Analytics", 40]
total = 0
i = 0

while i < len(items):
    if isinstance(items[i], int):
        total = total + items[i]

    i = i + 1

print("Sum of integers:", total)
```

```python
'''7.	Write a python program to count the number of strings from a given list of strings.  
The string length is 2 or more and the first and last characters are the same.
(hint: in string indexing is possible)
sample list: ['abc','xyz','aba','1221']
expected result:2'''

sample = ['abc', 'xyz', 'aba', '1221']

count = 0
i = 0

while i < len(sample):
    if len(sample[i]) >= 2 and sample[i][0] == sample[i][-1]:
        count = count + 1

    i = i + 1

print("Count:", count)
```
